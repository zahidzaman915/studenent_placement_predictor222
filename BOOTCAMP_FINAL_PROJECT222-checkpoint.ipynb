```python
import os
os.makedirs("src", exist_ok=True)
os.makedirs("models", exist_ok=True)     # needed later for the saved model
print("Working directory:", os.getcwd())
print(os.listdir("."))
```

```python
%%writefile src/features.py
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler

TECH_SKILLS = ["AWS", "Algorithms", "C++", "Cloud Computing", "Computer Vision", "Data Structures",
               "Data Visualization", "Deep Learning", "Docker", "Excel", "FastAPI", "Flask", "Git",
               "GitHub", "Java", "Machine Learning", "Matplotlib", "NLP", "NumPy", "OOP", "Pandas",
               "Power BI", "Python", "PyTorch", "SQL", "Scikit-learn", "Statistics", "TensorFlow"]
SOFT_SKILLS = ["Adaptability", "Collaboration", "Creativity", "Critical Thinking", "Decision Making",
               "Leadership", "Negotiation", "Presentation", "Teamwork", "Time Management"]
CERTIFICATES = ["AWS Certificate", "Cisco Certificate", "Coursera", "DigiSkills", "Google Certificate",
                "IBM Certificate", "Microsoft Certificate", "NAVTTC", "Udemy"]
LEVELS = ["Low", "Basic", "Good", "Very Good", "Excellent"]

LIST_COLS = {"technical_skills": (TECH_SKILLS, "tech", "tech_skill_count"),
             "soft_skills": (SOFT_SKILLS, "soft", "soft_skill_count"),
             "international_certificates": (CERTIFICATES, "cert", "cert_count")}

def _slug(t):
    return t.lower().replace("+", "plus").replace("-", "_").replace(" ", "_")

def _items(s):
    return set() if pd.isna(s) or s in ("", "None") else {t.strip() for t in s.split(",")}

BINARY_COLS = [f"{p}_{_slug(i)}" for cat, p, _ in LIST_COLS.values() for i in cat]
NUMERIC_COLS = ["age", "CGPA", "real_world_projects", "past_work_experience",
                "tech_skill_count", "soft_skill_count", "cert_count"]
ORDINAL_COLS = ["problem_solving_skills", "communication_skills"]
NOMINAL_COLS = ["gender", "extracurricular_activities", "club_membership"]

def engineer_features(X):
    """Stateless: turns list-like text columns into counts + 0/1 indicators."""
    X = X.copy()
    for col, (catalog, prefix, count_name) in LIST_COLS.items():
        sets = X[col].apply(_items)
        X[count_name] = sets.apply(len)
        for item in catalog:
            X[f"{prefix}_{_slug(item)}"] = sets.apply(lambda s, i=item: int(i in s))
    return X.drop(columns=list(LIST_COLS))

def build_preprocessor(scale=False):
    num_steps = [("imputer", SimpleImputer(strategy="median"))]
    ord_steps = [("imputer", SimpleImputer(strategy="most_frequent")),
                 ("encoder", OrdinalEncoder(categories=[LEVELS, LEVELS],
                                            handle_unknown="use_encoded_value", unknown_value=-1))]
    if scale:
        num_steps.append(("scaler", StandardScaler()))
        ord_steps.append(("scaler", StandardScaler()))
    return ColumnTransformer([
        ("num", Pipeline(num_steps), NUMERIC_COLS),
        ("ord", Pipeline(ord_steps), ORDINAL_COLS),
        ("nom", Pipeline([("imputer", SimpleImputer(strategy="constant", fill_value="None")),
                          ("encoder", OneHotEncoder(handle_unknown="ignore"))]), NOMINAL_COLS),
        ("bin", "passthrough", BINARY_COLS)])
```

# Phase 7: Feature Selection
## 7.0 Add a Helper to Drop Columns from the Pipeline

```python
%%writefile -a src/features.py


def build_subset_preprocessor(scale=False, exclude=()):
    """Same as build_preprocessor but drops the listed columns (raw or engineered)."""
    prep = build_preprocessor(scale=scale)
    new = []
    for name, trans, cols in prep.transformers:
        cols = [c for c in cols if c not in exclude]
        if name == "ord" and cols:
            trans.named_steps["encoder"].set_params(categories=[LEVELS] * len(cols))
        new.append((name, trans, cols))
    prep.set_params(transformers=new)
    return prep
```

## Phase 8: Tuning, Final Evaluation and Saving
## 8.1 Hyperparameter Tuning (GridSearchCV, Training Set Only)

```python
from sklearn.model_selection import GridSearchCV, cross_val_predict
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
                             average_precision_score, brier_score_loss, ConfusionMatrixDisplay)

FEATURES = ["CGPA", "technical_skills", "soft_skills", "international_certificates",
            "extracurricular_activities", "club_membership", "real_world_projects",
            "past_work_experience", "problem_solving_skills", "communication_skills"]
EXCL = ("gender", "age") + tuple(BINARY_COLS)
X_tr, X_te = X_train[FEATURES], X_test[FEATURES]

grid_lr = GridSearchCV(
    make_pipe_sub(LogisticRegression(max_iter=1000, random_state=42), True, EXCL),
    {"model__C": [0.01, 0.03, 0.1, 0.3, 1, 10]}, cv=cv,
    scoring={"neg_log_loss": "neg_log_loss", "pr_auc": "average_precision", "roc_auc": "roc_auc"},
    refit="neg_log_loss", n_jobs=-1).fit(X_tr, y_train)

grid_dt = GridSearchCV(
    make_pipe_sub(DecisionTreeClassifier(random_state=42), False, EXCL),
    {"model__max_depth": [3, 4, 5, 6, 8], "model__min_samples_leaf": [10, 20, 50],
     "model__class_weight": [None, "balanced"]},
    cv=cv, scoring="f1", n_jobs=-1).fit(X_tr, y_train)

print("LR best:", grid_lr.best_params_)
print("DT best:", grid_dt.best_params_, "| CV F1:", round(grid_dt.best_score_, 3))
```

### 8.2 Decision Threshold Selection via Cross-Validation

```python
best_lr = grid_lr.best_estimator_
proba_cv = cross_val_predict(best_lr, X_tr, y_train, cv=cv, method="predict_proba")[:, 1]

thresholds = np.arange(0.20, 0.65, 0.01)
f1s = [f1_score(y_train, proba_cv >= t) for t in thresholds]
THRESHOLD = round(float(thresholds[int(np.argmax(f1s))]), 2)
print("Selected threshold:", THRESHOLD, "| CV F1 at this threshold:", round(max(f1s), 3))
```

```python
rows = []
def evaluate(name, model, thr):
    p = model.predict_proba(X_te)[:, 1]; yp = (p >= thr).astype(int)
    rows.append({"Model": name, "Accuracy": accuracy_score(y_test, yp), "Precision": precision_score(y_test, yp),
                 "Recall": recall_score(y_test, yp), "F1": f1_score(y_test, yp),
                 "ROC-AUC": roc_auc_score(y_test, p), "PR-AUC": average_precision_score(y_test, p)})
    return yp

evaluate("Decision Tree (tuned, thr 0.5)", grid_dt.best_estimator_, 0.5)
evaluate("Logistic Regression (tuned, thr 0.5)", best_lr, 0.5)
y_pred = evaluate(f"Logistic Regression (tuned, thr {THRESHOLD})", best_lr, THRESHOLD)

display(pd.DataFrame(rows).set_index("Model").round(3))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=["Not placed", "Placed"], cmap="Blues")
plt.title("Final Model: Confusion Matrix (Test Set)"); plt.show()
```

#### 8.5 Prediction Module (src/predict.py)

```python
%%writefile src/predict.py
import json
from pathlib import Path

import joblib
import pandas as pd

from src.features import CERTIFICATES, LEVELS, SOFT_SKILLS, TECH_SKILLS

ROOT = Path(__file__).resolve().parent.parent
MODEL_PATH = ROOT / "models" / "placement_model_v1.joblib"
META_PATH = ROOT / "models" / "model_metadata.json"

FEATURES = ["CGPA", "technical_skills", "soft_skills", "international_certificates",
            "extracurricular_activities", "club_membership", "real_world_projects",
            "past_work_experience", "problem_solving_skills", "communication_skills"]


def load_model():
    model = joblib.load(MODEL_PATH)
    meta = json.loads(META_PATH.read_text())
    return model, meta


def _validate(p):
    missing = [f for f in FEATURES if f not in p]
    if missing:
        raise ValueError(f"Missing inputs: {missing}")
    if not 0.0 <= float(p["CGPA"]) <= 4.0:
        raise ValueError("CGPA must be between 0.0 and 4.0")
    if not 0 <= int(p["real_world_projects"]) <= 5:
        raise ValueError("Projects must be between 0 and 5 (range seen in training data)")
    if not 0 <= int(p["past_work_experience"]) <= 4:
        raise ValueError("Work experience must be between 0 and 4 (range seen in training data)")
    for col in ("problem_solving_skills", "communication_skills"):
        if p[col] not in LEVELS:
            raise ValueError(f"{col} must be one of {LEVELS}")
    for col, catalog in (("technical_skills", TECH_SKILLS), ("soft_skills", SOFT_SKILLS),
                         ("international_certificates", CERTIFICATES)):
        bad = [i for i in p[col] if i not in catalog]
        if bad:
            raise ValueError(f"Unknown items in {col}: {bad}")


def predict_placement(profile, model=None, meta=None):
    """profile: dict with the 10 FEATURES (list-type fields are Python lists)."""
    if model is None or meta is None:
        model, meta = load_model()
    _validate(profile)
    row = {k: profile[k] for k in FEATURES}
    for col in ("technical_skills", "soft_skills", "international_certificates"):
        row[col] = ", ".join(row[col]) if row[col] else "None"
    X = pd.DataFrame([row])
    prob = float(model.predict_proba(X)[0, 1])
    placed = prob >= meta["threshold"]
    return {"placed": bool(placed),
            "label": "Likely to be Placed" if placed else "Lower Likelihood of Placement",
            "probability": round(prob, 4),
            "threshold": meta["threshold"],
            "note": "Model estimate based on the entered profile; not a guarantee of any outcome."}
```

### 9.3 Personalised Recommendation Engine (src/recommend.py)

```python
%%writefile src/recommend.py
import math

from src.features import LEVELS
from src.predict import predict_placement

TECH_PRIORITY = ["Python", "SQL", "Machine Learning", "Git", "Pandas", "Data Visualization",
                 "Statistics", "Docker", "Cloud Computing"]
CERT_PRIORITY = ["Google Certificate", "Coursera", "Microsoft Certificate", "IBM Certificate",
                 "AWS Certificate", "Udemy"]
SOFT_PRIORITY = ["Leadership", "Presentation", "Critical Thinking", "Teamwork", "Time Management"]


def _logit(p):
    p = min(max(p, 1e-6), 1 - 1e-6)
    return math.log(p / (1 - p))


def _next_level(level):
    return LEVELS[min(LEVELS.index(level) + 1, len(LEVELS) - 1)]


def _missing(owned, priority, k):
    return [x for x in priority if x not in owned][:k]


def improvement_levers(profile, model, meta, top_n=4):
    """What-if analysis: apply one realistic improvement at a time, re-score it with the
    model, and return the actions with the largest estimated gain for THIS profile."""
    base = _logit(predict_placement(profile, model, meta)["probability_raw"])
    found = []

    def add(title, detail, changes):
        gain = _logit(predict_placement({**profile, **changes}, model, meta)["probability_raw"]) - base
        if gain > 0.02:
            impact = "High impact" if gain >= 0.5 else "Medium impact" if gain >= 0.25 else "Helpful"
            found.append({"title": title, "detail": detail, "gain": gain, "impact": impact})

    cg = float(profile["CGPA"])
    if cg < 3.8:
        tgt = round(min(4.0, cg + 0.3), 2)
        add("Raise your CGPA", f"Work toward about {tgt:.2f} through consistent study and by "
            "strengthening your weaker courses.", {"CGPA": tgt})

    n = int(profile["real_world_projects"])
    if n < 5:
        add("Build more real-world projects", f"You listed {n}. Add a practical project (data analysis, "
            "machine learning or a web app) and publish it on GitHub.", {"real_world_projects": n + 1})

    tech = list(profile["technical_skills"])
    if len(tech) < 8:
        new = _missing(tech, TECH_PRIORITY, min(2, 8 - len(tech)))
        if new:
            add("Expand your technical skills", f"You listed {len(tech)}. Consider learning "
                f"{' and '.join(new)}.", {"technical_skills": tech + new})

    e = int(profile["past_work_experience"])
    if e < 4:
        add("Gain practical work experience", f"You listed {e}. Look for internships, freelance work, "
            "part-time roles or open-source contributions.", {"past_work_experience": e + 1})

    certs = list(profile["international_certificates"])
    if len(certs) < 3:
        new = _missing(certs, CERT_PRIORITY, 1)
        if new:
            add("Earn a recognised certification", f"You listed {len(certs)}. A certificate such as "
                f"{new[0]} would strengthen your profile.", {"international_certificates": certs + new})

    soft = list(profile["soft_skills"])
    if len(soft) < 6:
        new = _missing(soft, SOFT_PRIORITY, 1)
        if new:
            add("Develop your soft skills", f"You listed {len(soft)}. Build {new[0].lower()} through "
                "team projects and student societies.", {"soft_skills": soft + new})

    if profile["communication_skills"] != LEVELS[-1]:
        add("Strengthen communication", "Practise presentations, join debate or speaking sessions, "
            "and rehearse interview answers.",
            {"communication_skills": _next_level(profile["communication_skills"])})

    if profile["problem_solving_skills"] != LEVELS[-1]:
        add("Sharpen problem-solving", "Solve coding and logic problems regularly "
            "(algorithms, data structures, case studies).",
            {"problem_solving_skills": _next_level(profile["problem_solving_skills"])})

    if profile["club_membership"] == "None":
        add("Join a technical club", "Active club membership shows initiative and builds your network.",
            {"club_membership": "DSCS - Member"})

    if profile["extracurricular_activities"] == "None":
        add("Take part in activities", "Hackathons, workshops or volunteering add visible experience.",
            {"extracurricular_activities": "Hackathon"})

    return sorted(found, key=lambda d: d["gain"], reverse=True)[:top_n]


def profile_strengths(profile):
    s = []
    if float(profile["CGPA"]) >= 3.3:
        s.append("Strong academic record (CGPA)")
    if len(profile["technical_skills"]) >= 6:
        s.append("Broad technical skill set")
    if int(profile["real_world_projects"]) >= 3:
        s.append("Solid portfolio of real-world projects")
    if int(profile["past_work_experience"]) >= 2:
        s.append("Relevant work experience")
    if len(profile["international_certificates"]) >= 2:
        s.append("Multiple professional certifications")
    if profile["problem_solving_skills"] in LEVELS[-2:]:
        s.append("Strong problem-solving ability")
    return s
```

### 9.4 Streamlit Application (app/streamlit_app.py)

```python
%%writefile app/streamlit_app.py
import logging
import sys
from pathlib import Path

import streamlit as st

ROOT = Path(__file__).resolve().parent.parent
sys.path.insert(0, str(ROOT))

from src.features import CERTIFICATES, LEVELS, SOFT_SKILLS, TECH_SKILLS  # noqa: E402
from src.predict import load_model, predict_placement  # noqa: E402
from src.recommend import improvement_levers, profile_strengths  # noqa: E402

logging.basicConfig(level=logging.INFO)
log = logging.getLogger("placement_app")

st.set_page_config(page_title="Student Placement Predictor", page_icon="🎓", layout="wide")

CLUBS = ["None", "DSCS - Member", "DSCS - Active Member", "DSCS - Event Organizer", "DSCS - Executive",
         "University Tech Club - Member", "University Tech Club - Executive"]
ACTIVITIES = ["None", "Community Service", "Debate", "Hackathon", "Research Activity", "Sports",
              "Tech Event Organizer", "Volunteer Work", "Workshop Organizer"]

st.markdown("""
<style>
.block-container {padding-top: 1.5rem; max-width: 1150px;}
.hero {background: linear-gradient(120deg, #12294d 0%, #1F4E8C 60%, #2f7bd1 100%);
       padding: 2rem 2.2rem; border-radius: 16px; color: #fff; margin-bottom: 1.2rem;}
.hero h1 {color: #fff; margin: 0 0 .3rem 0; font-size: 2.1rem;}
.hero p {color: #dce8f7; margin: 0; font-size: 1.02rem;}
.sec {font-weight: 700; color: #12294d; font-size: 1.05rem; margin-bottom: .4rem;}
.result {padding: 1.4rem 1.6rem; border-radius: 14px; margin: .8rem 0 1rem 0; color: #1B2430;}
.result h2 {margin: 0 0 .5rem 0;}
.result.good {background: #E8F5E9; border-left: 8px solid #2E7D32;}
.result.low {background: #EAF2FB; border-left: 8px solid #1F4E8C;}
.tip {background: #fff; border: 1px solid #dfe6f1; border-radius: 12px; padding: .8rem 1rem;
      margin-bottom: .6rem; color: #1B2430;}
.tip b {color: #12294d;}
.gain {color: #2E7D32; font-weight: 600; font-size: .9rem;}
.small {color: #5b6676; font-size: .85rem;}
</style>
""", unsafe_allow_html=True)


def fmt_pct(p):
    return "<1%" if p < 0.01 else ">99%" if p > 0.99 else f"{p:.0%}"


@st.cache_resource
def get_model():
    return load_model()


try:
    model, meta = get_model()
except Exception as exc:  # pragma: no cover
    st.error("The prediction model could not be loaded. Please contact the project owner.")
    log.exception("Model load failed: %s", exc)
    st.stop()

st.markdown("""
<div class="hero">
  <h1>🎓 Student Placement Predictor</h1>
  <p>Estimate placement likelihood from academic, technical and professional profile factors,
  and get personalised guidance on how to strengthen it.</p>
</div>
""", unsafe_allow_html=True)

with st.sidebar:
    st.header("About the Project")
    st.write("A machine-learning system that predicts whether a student is likely to be placed, "
             "trained on 5,000 synthetic student records.")
    tm = meta["test_metrics"]
    st.markdown("**Model:** Logistic Regression  \n"
                f"**Recall (placed students):** {tm['recall']:.0%}  \n"
                f"**Precision:** {tm['precision']:.0%}  \n"
                f"**ROC-AUC:** {tm['roc_auc']:.2f}")
    st.markdown("**Privacy:** no name, CNIC, gender or age is collected or used.")
    st.caption("Results are statistical estimates for guidance only. They are not a guarantee "
               "of any hiring or placement outcome.")

with st.form("profile"):
    c1, c2 = st.columns(2, gap="large")
    with c1:
        with st.container(border=True):
            st.markdown('<div class="sec">1 · Student Academic Profile</div>', unsafe_allow_html=True)
            cgpa = st.slider("CGPA (out of 4.0)", 0.0, 4.0, 3.0, 0.01, key="cgpa")
        with st.container(border=True):
            st.markdown('<div class="sec">2 · Technical Skills</div>', unsafe_allow_html=True)
            tech = st.multiselect("Select your technical skills", TECH_SKILLS, key="tech")
        with st.container(border=True):
            st.markdown('<div class="sec">3 · Soft Skills</div>', unsafe_allow_html=True)
            soft = st.multiselect("Select your soft skills", SOFT_SKILLS, key="soft")
        with st.container(border=True):
            st.markdown('<div class="sec">4 · Certifications</div>', unsafe_allow_html=True)
            certs = st.multiselect("Certificates earned", CERTIFICATES, key="certs")
    with c2:
        with st.container(border=True):
            st.markdown('<div class="sec">5 · Projects</div>', unsafe_allow_html=True)
            projects = st.slider("Real-world projects completed", 0, 5, 1, key="projects")
        with st.container(border=True):
            st.markdown('<div class="sec">6 · Work Experience</div>', unsafe_allow_html=True)
            exp = st.slider("Internships / jobs / freelance experiences", 0, 4, 0, key="exp")
        with st.container(border=True):
            st.markdown('<div class="sec">7 · Problem-Solving Skills</div>', unsafe_allow_html=True)
            ps = st.select_slider("Your level", LEVELS, value="Good", key="ps")
        with st.container(border=True):
            st.markdown('<div class="sec">8 · Communication Skills</div>', unsafe_allow_html=True)
            comm = st.select_slider("Your level", LEVELS, value="Good", key="comm")
        with st.container(border=True):
            st.markdown('<div class="sec">Activities and Clubs</div>', unsafe_allow_html=True)
            club = st.selectbox("Club membership", CLUBS, key="club")
            act = st.selectbox("Extracurricular activity", ACTIVITIES, key="act")
    submitted = st.form_submit_button("Predict Placement Likelihood", type="primary",
                                      use_container_width=True)

if submitted:
    profile = {"CGPA": cgpa, "technical_skills": tech, "soft_skills": soft,
               "international_certificates": certs, "extracurricular_activities": act,
               "club_membership": club, "real_world_projects": projects,
               "past_work_experience": exp, "problem_solving_skills": ps,
               "communication_skills": comm}
    try:
        res = predict_placement(profile, model, meta)
    except ValueError as err:
        st.error(f"Please check your inputs: {err}")
        st.stop()
    log.info("prediction made: placed=%s prob=%.3f", res["placed"], res["probability"])
    prob = res["probability"]

    st.markdown("---")
    st.subheader("Prediction Result")
    if res["placed"]:
        st.markdown("""
<div class="result good"><h2>🎉 Congratulations!</h2>
<p>Based on your academic performance, technical skills, projects, experience and other relevant
factors, the model predicts a <b>strong likelihood of placement</b> for your profile.</p></div>
""", unsafe_allow_html=True)
    else:
        near = (meta["threshold"] - prob) <= 0.10
        extra = " You are close to the positive range, so a few targeted improvements could make a difference." if near else ""
        st.markdown(f"""
<div class="result low"><h2>Your Placement Outlook</h2>
<p>Thank you for completing the assessment. Based on the current profile, the model predicts a
<b>lower likelihood of placement at this stage</b>. This is a snapshot of today's profile, not a
verdict on your potential.{extra}</p></div>
""", unsafe_allow_html=True)

    m1, m2 = st.columns([1, 2])
    m1.metric("Estimated placement likelihood", fmt_pct(prob))
    with m2:
        st.progress(min(max(prob, 0.0), 1.0))
        st.caption(f"The model classifies a profile as 'likely placed' at or above "
                   f"{meta['threshold']:.0%}. This is an estimate, not a guarantee.")

    if res["placed"]:
        strengths = profile_strengths(profile)
        if strengths:
            st.markdown("#### What is working in your favour")
            for s in strengths:
                st.markdown(f"- {s}")
        levers = improvement_levers(profile, model, meta, top_n=2)
        if levers:
            st.markdown("#### To stay competitive")
            for lv in levers:
                st.markdown(f'<div class="tip"><b>{lv["title"]}</b><br>{lv["detail"]}</div>',
                            unsafe_allow_html=True)
    else:
        levers = improvement_levers(profile, model, meta, top_n=4)
        st.markdown("#### Recommended next steps for your profile")
        if levers:
            for lv in levers:
                st.markdown(
                    f'<div class="tip"><b>{lv["title"]}</b> '
                    f'<span class="gain">({lv["impact"]})</span>'
                    f'<br>{lv["detail"]}</div>', unsafe_allow_html=True)
        else:
            st.info("Your profile is already strong across the factors the model considers.")

    with st.expander("How to read this result"):
        st.markdown(
            "- The percentage is the model's estimated likelihood, based on patterns in 5,000 "
            "synthetic student records.\n"
            "- Recommendations come from re-scoring your profile with one improvement at a time. "
            "The impact label reflects how much the model's estimate would move. It shows association, not proven cause and effect.\n"
            "- Real hiring depends on many factors this model cannot see.")
```
